```python
%sql
-- Código SQL para criar o Volume no Unity Catalog

-- 1. Cria o Catálogo principal, se ele ainda não existir
CREATE CATALOG IF NOT EXISTS bigdata_anomaly_detection_kddcup99_catalogue;

-- 2. Seleciona o Catálogo para garantir que o Schema e o Volume serão criados nele
USE CATALOG bigdata_anomaly_detection_kddcup99_catalogue;

-- 3. Cria o Schema (Banco de Dados) 'default', se ele ainda não existir
CREATE SCHEMA IF NOT EXISTS default;

-- 4. Cria o Volume que será usado para armazenar o dataset raw (kddcup99) e artefatos temporários
-- O caminho físico será: /Volumes/bigdata_anomaly_detection_kddcup99_catalogue/default/kdd_volume
CREATE VOLUME IF NOT EXISTS default.kdd_volume
COMMENT 'Volume para armazenar o dataset KDDCup99 raw e artefatos temporários do MLflow.';
```
